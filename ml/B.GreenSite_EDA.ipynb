cell id

```python
plt.figure(figsize = (10,6))

plt.hist(df['cell_id'], bins = 50)
plt.title('cell_id')
plt.xlabel('cell_id')
plt.ylabel('Frequancy')

mean = df['cell_id'].mean()
median = df['cell_id'].median()

plt.axvline(mean, color = 'g', label = 'Mean', linestyle = '--')
plt.axvline(median, color = 'orange', label = 'Median', linestyle = '--')

plt.legend()
plt.grid()
plt.show()
```

```python
plt.boxplot(df['cell_id'])
plt.grid()
plt.show()
```

ndvi summary qa

```python
plt.figure(figsize = (10,6))

plt.hist(df['ndvi_summary_qa'], bins = 50)
plt.title('ndvi_summary_qa')
plt.xlabel('ndvi_summary_qa')
plt.ylabel('Frequancy')

mean = df['ndvi_summary_qa'].mean()
median = df['ndvi_summary_qa'].median()

plt.axvline(mean, color = 'g', label = 'Mean', linestyle = '--')
plt.axvline(median, color = 'orange', label = 'Median', linestyle = '--')

plt.legend()
plt.grid()
plt.show()
```

*most of the data are correct

```python
plt.boxplot(df['ndvi_summary_qa'])
plt.grid()
plt.show()
```

```python
Q1 = df['ndvi_summary_qa'].quantile(0.25)
Q3 = df['ndvi_summary_qa'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

df[(df['ndvi_summary_qa'] < lower) | (df['ndvi_summary_qa'] > upper)].shape
```

```python
df[df['ndvi_summary_qa'] > upper].describe()
```

ndvi raw

```python
plt.figure(figsize = (10,6))

plt.hist(df['ndvi_raw'], bins = 50)
plt.title('ndvi_raw')
plt.xlabel('ndvi_raw')
plt.ylabel('Frequancy')

mean = df['ndvi_raw'].mean()
median = df['ndvi_raw'].median()

plt.axvline(mean, color = 'g', label = 'Mean', linestyle = '--')
plt.axvline(median, color = 'orange', label = 'Median', linestyle = '--')

plt.legend()
plt.grid()
plt.show()
```

```python
plt.boxplot(df['ndvi_raw'])
plt.grid()
plt.show()
```

```python
Q1 = df['ndvi_raw'].quantile(0.25)
Q3 = df['ndvi_raw'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

df[(df['ndvi_raw'] < lower) | (df['ndvi_raw'] > upper)].shape
```

```python
df[df['ndvi_raw'] < lower].describe()
```

```python
plt.figure(figsize = (10,6))

my_sc = plt.scatter(df['latitude'] , df['longitude'], c = df['ndvi_raw'], cmap = 'YlGn',vmax=10000, alpha = 0.2)
my_bar = plt.colorbar(my_sc)
my_bar.set_label('ndvi')

plt.title('Vegetation Of Baghdad City')
plt.xlabel('latitude')
plt.ylabel('longitude')
plt.show()
```
